```python
# 1 Escribe una función que reciba una cadena de texto como parámetro y devuelva un diccionario con las frecuencias 
#de cada letra en la cadena. Los espacios no deben ser considerados.


chain = "Master of Puppets"

def frequencyRecount(chain):
    frequencies = {}
    for letter in chain:
        if letter == " ": #En este if solamente pedimos que se ignoren los espacios, si la cadena los tiene.
            continue
        if letter in frequencies: #Si la letra ya está incluida en las frecuencias, se le suma 1 a su valor
            frequencies[letter] +=1
        else:                        #Y si no, se crea y toma el valor 1
            frequencies[letter] = 1

    return frequencies

print(frequencyRecount(chain))
```

```python
# 30 Crea una función que determine si dos palabras son anagramas, es decir, si están formadas por las mismas letras 
# pero en diferente orden.

def checkAnagram(word1, word2):
    word1low = word1.replace(" ", "").lower() #En un solo paso, convertimos cada palabra a minúsculas para facilitar la comparativa, y eliminamos los posibles espacios.
    word2low = word2.replace(" ", "").lower()
    
    #El método sorted() va perfecto aquí, para comparar ambas palabras con sus letras ordenadas alfabéticamente.
    result = sorted(word1low) == sorted(word2low) #Result será un boolean, True o False.
    return result

checkAnagram("Trueno", "Neutro")
```

```python
# 34 Crea la clase Arbol
#Define un árbol genérico con un tronco y ramas como atributos.
#Métodos disponibles: crecer_tronco, nueva_rama, crecer_ramas, quitar_rama, info_arbol.
#Código a seguir:
#Inicializar un árbol con un tronco de longitud 1 y una lista vacía de ramas.
#Implementar el método crecer_tronco para aumentar la longitud del tronco en una unidad.
#Implementar el método nueva_rama para agregar una nueva rama de longitud 1 a la lista de ramas.
#Implementar el método crecer_ramas para aumentar en una unidad la longitud de todas las ramas existentes.
#Implementar el método quitar_rama para eliminar una rama en una posición específica.
#Implementar el método info_arbol para devolver información sobre la longitud del tronco, el número de ramas y sus longitudes.
#Caso de uso:
        #a. Crear un árbol.
        #b. Hacer crecer el tronco una unidad.
        #c. Añadir una nueva rama.
        #d. Hacer crecer todas las ramas una unidad.
        #e. Añadir dos nuevas ramas.
        #f. Retirar la rama situada en la posición 2.
        #g. Obtener información sobre el árbol.

#Empieza la parte de OOP del temario.        
class Arbol:    
        def __init__(self): #Esta es la función de inicio para definir la clase árbol. El "self" se ha puesto solo, imagino que se pone así como costumbre.
                self.tronco = 1 #Inicializamor un tronco con longitud 1
                self.ramas = [] # Y una lista vacía de ramas
                
        def crecer_tronco(self): #Cada vez que se llame a este método, el tronco crecerá una unidad
                self.tronco += 1
                
        def nueva_rama(self): #Con este método añadimos una rama a la lista cada vez.
                self.ramas.append(1) 
                
        def crecer_ramas(self):
                self.ramas = [rama +1 for rama in self.ramas] #Sumamos 1 al valor de cada rama que ya haya en la lista con un bucle for acotado. La sintaxis de esto es casi como leerlo en idioma humano.
                
        def quitar_rama(self, posicionRama):
                if posicionRama > len(self.ramas): #Si el número de posición de la rama que queremos eliminar es inferior a la longitud de la lista de ramas, se elimina esa posición usando pop. Recordatorio: 0 es la primera pos., 1 la segunda, etc.
                      self.ramas.pop(posicionRama)  
                      
        def info_arbol(self): #Con esta función sacamos la info actual del árbol por consola.
                print("El tronco mide ", self.tronco, ".")
                print("Hay ", len(self.ramas), " ramas, que miden ", self.ramas, ".")
                
#Y empezamos el caso de uso, asociando primero una variable a la clase Arbol.

arbol = Arbol()    #a) He tenido que consultar la IA porque me faltaban los paréntesis.
      
arbol.crecer_tronco()   #b)  
arbol.nueva_rama() #c)
arbol.crecer_ramas() #d)

for _ in range(2): #e) Me he buscado este bucle para no tener que escribir dos veces el mismo método para añadir dos ramas. 
        arbol.nueva_rama() 

arbol.quitar_rama(2)  #f)
arbol.info_arbol()  #g)
        
        
```

```python
# 35 Crea la clase UsuarioBanco
#Representa a un usuario de un banco con su nombre, saldo y si tiene o no cuenta corriente.
#Métodos: retirar_dinero, transferir_dinero, agregar_dinero.
#Código a seguir:
#Inicializar un usuario con nombre, saldo y un indicador (True o False) de cuenta corriente.
#Implementar retirar_dinero para sustraer dinero del saldo, lanzando un error si no es posible.
#Implementar transferir_dinero para transferir dinero desde otro usuario, lanzando un error en caso de fallo.
#Implementar agregar_dinero para aumentar el saldo del usuario.
#Caso de uso:
        #a. Crear dos usuarios: "Alicia" con saldo inicial de 100 y "Bob" con saldo inicial de 50, ambos con cuenta corriente.
        #b. Agregar 20 unidades al saldo de Bob.
        #c. Transferir 80 unidades de Bob a Alicia.
        #d. Retirar 50 unidades del saldo de Alicia.
        
        
class UsuarioBanco: #Iniciamos la clase
    def __init__(self, nombre, saldo, tieneCuenta = True): #Función de inicio con sus atributos: su entidad propia, nombre, saldo y tenencia de cuenta (valor booleano).
        self.nombre = nombre
        self.saldo = saldo
        self.tieneCuenta = tieneCuenta
        
    def retirar_dinero(self, cantidad): #La función para retirar dinero. Hay que meterle por atributo la cantidad a retirar.
        if cantidad <= 0: #Error si el usuario introduce un valor negativo.
            raise ValueError("Error, introduzca una cantidad válida.")
        if cantidad > self.saldo: #Error si solicita retirar más dinero del que tiene en la cuenta.
            raise ValueError("Error, saldo insuficiente.")
        self.saldo -= cantidad #Si ninguno de los errores salta, realizamos la operación.
        print(f"Operación realizada correctamente. Su saldo actual es de {self.saldo}.") #Imprimimos el saldo restante.
        
    def transferir_dinero(self, otroUsuario, cantidad):
        if self.tieneCuenta and otroUsuario.tieneCuenta: #Si los dos usuarios tienen cuenta, lanzamos el resto de errores u operaciones con bucles anidados. Si no, saltamos al else principal.
            if cantidad <= 0:
                raise ValueError("Error, introduzca una cantidad válida.") #Mismos posibles errores que en el método anterior.
            if cantidad > self.saldo:
                raise ValueError("Error, saldo insuficiente.")
            self.saldo -= cantidad
            otroUsuario.saldo += cantidad
            print(f"Transferencia realizada correctamente. Su saldo actual es de {self.saldo}.") #No sacamos el saldo del otro usuario porque sería difundir su intimidad jajaja   
        else:
            raise ValueError("Error, ambos usuarios deben poseer una cuenta corriente.")
        
    def agregar_dinero(self, cantidad):
        if cantidad <= 0: #Mismo error por cantidad negativa o 0 que en los métodos anteriores. En este caso no es necesario el error por falta de saldo, al tratarse de un ingreso.
            raise ValueError("Error, introduzca una cantidad válida.")
        self.saldo += cantidad
        print(f"Operación realizada correctamente. Su saldo actual es de {self.saldo}.")
            
#Vamos al caso de uso:
alicia = UsuarioBanco("Alicia", 100, True) #a) Creamos una variable por cada usuario, con sus datos como atributos.
bob = UsuarioBanco("Bob", 50, True)

bob.agregar_dinero(20)

bob.transferir_dinero(alicia, 80) # Esto dará error, porque el saldo de Bob es inferior a la transferencia que quiere hacer.

alicia.retirar_dinero(50)

#Si lo he entendido bien, el error que sale por consola es el esperable. Si anulo la línea de la transferencia de bob, las otras dos operaciones se realizan sin problema.
        
```

```python
# 39 Escribe una función que tome dos parámetros: figura (una cadena que puede ser "rectangulo", "circulo" o "triangulo")
# y datos (una tupla con los datos necesarios para calcular el área de la figura).

#Me voy a complicar un poco la vida, porque he buscado cómo eliminar las tildes de las figuras recibidas por argumento, 
#para evitar errores e incoincidencias. Lo hago con una función extra e importando unicodedata:

import unicodedata

def estandarizarTexto(figura):
    figura = figura.lower() #Primero pasamos la figura a minúscula.
    figura = "".join(
        c for c in unicodedata.normalize("NFD", figura) #Esto me lo ha dado la IA.
        if unicodedata.category(c) != "Mn"
    )
    return figura 

def calcularArea(figura, *args):
    figura = estandarizarTexto(figura) #Usamos la función anterior para tener el nombre de la figura sin tildes ni mayúsculas.
    
    if figura == "rectangulo":
        if len(args) != 2:
            raise ValueError("Error: para calcular el área del rectángulo necesitamos su base y su altura.")
        base, altura = args
        return base * altura
    
    elif figura == "circulo":
        if len(args) != 1:
            raise ValueError("Error: solo necesitamos el radio del círculo.")
        (radio, ) = args #Esto me estaba dando problemas, porque no estaba 'desempaquetando' la tupla en el caso del círculo. Como es un único valor, hay que explicarlo con paréntesis y la coma (me lo ha chivado la IA).
        import math #Hay que importar el módulo de matemáticas para hacer esta operación.
        return math.pi * radio**2
    
    elif figura == "triangulo":
        if len(args) != 2:
            raise ValueError("Error: necesitamos la base y altura del triángulo.")
        base, altura = args
        return (base * altura) / 2
    
    else:
        raise ValueError("Error: figura no reconocida. Usa 'rectangulo', 'circulo' o 'triangulo'.")
            
calcularArea("circulo", 53)
    
    
    
```

```python
# 40 Escribe un programa en Python que utilice condicionales para determinar el monto final de una compra en una tienda en
# línea, después de aplicar un descuento. El programa debe:
    #a. Solicitar al usuario el precio original de un artículo.
    #b. Preguntar si tiene un cupón de descuento (respuesta sí o no).
    #c. Si la respuesta es sí, solicitar el valor del cupón de descuento.
    #d. Aplicar el descuento al precio original, siempre que el valor del cupón sea válido (mayor a cero).
    #e. Mostrar el precio final de la compra, considerando o no el descuento.
    #f. Usar estructuras de control de flujo (if, elif, else) para llevar a cabo las acciones.
    
def calcularPrecioTotal():
    #Van a ser un montón de condicionales anidados, así que los voy a meter en una estructura de try-except, por darle variedad y manejar mejor el posible ValueError del precio.
    try:
        precioOriginal = float(input("Introduce el precio original del artículo:")) #Lo pasamos a float en lugar de int, porque probablemente el precio tenga decimales.
        tieneCupon = input("¿Tienes un cupón de descuento?").strip().lower() #La respuesta será 'Sí' o 'No'. Le quitamos las mayúsculas y posibles espacios.
        if tieneCupon == "si" or tieneCupon == "sí": #En el ejercicio anterior hice una función para eliminar tildes, pero en este caso es mucho más sencillo aceptar 'si' o 'sí' indistintamente. Si hubiese más casos de tildes en inputs, haría la función del ej anterior sin problema.
            cuponValor = float(input("Introduce el valor del cupón de descuento: "))
            if cuponValor > 0:
                precioFinal = precioOriginal - cuponValor
                if precioFinal < 0: #No podemos dar un precio negativo si el descuento es mayor al precio, así que el mínimo será 0.
                    precioFinal = 0
                print(f"El precio final tras aplicar el descuento es de {precioFinal}")
            else:
                print(f"Cupón no válido. El precio final es {precioOriginal}") #Si el cupón no es válido, el precio original no se modifica.
        elif tieneCupon == "no":
            print(f"El precio final es de {precioOriginal}")        
        else:
            print("Respuesta no válida. Por favor, responde 'sí' o 'no'.")  
    except ValueError:
        print("Error: introduce un precio válido.")     
        
calcularPrecioTotal()         
```
